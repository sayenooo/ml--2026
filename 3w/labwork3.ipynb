Q1: Loading and Understanding the Dataset (10 marks)

```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

```python
df = sns.load_dataset("tips")
df.head()
df.shape
df.columns
```

Q1: Loading and Understanding the Dataset (10 marks)

```python
weekend_cond1 = (df["day"]=="Sat") | (df["day"]=="Sun")
weekday_cond2 = (df["day"]!="Sat") & (df["day"]!="Sun")

weekend = df[weekend_cond1]["total_bill"].mean()
weekday = df[weekday_cond2]["total_bill"].mean()
overall = df["total_bill"].mean()

print(weekend)
print(weekday)
print(overall)


weekend = df[weekend_cond1]["tip"].mean()
weekday = df[weekday_cond2]["tip"].mean()
overall = df["tip"].mean()
print("--------------------------------")
print(weekend)
print(weekday)
print(overall)
print("--------------------------------")
print("Weekend customers have higher total bills and give higher tips on average compared to weekday customers.")
```

Q5: Visualization – What Influences Tips? (25 marks

```python
sns.scatterplot(x = "total_bill", y = "tip", data = df)

plt.xlabel("total_bill")
plt.ylabel("tip")
```

```python
sns.barplot(data = df, x = "day", y = "tip", estimator = "mean")
```

```python
sns.histplot(data = df, x = "tip", bins = 30, kde = True)
```

Customers:
Customers generally leave higher tips when the total bill is higher. Weekend customers,

especially on Saturday and Sunday, tend to tip more than weekday customers. Tips are also

higher during dinner time, which often involves larger groups and higher spending.


Restaurant Managers:
These findings show that weekends and dinner hours are important for restaurant revenue.

Managers can schedule more staff during these busy periods to provide better service.

Focusing on weekend customers and larger groups can help increase both sales and tips.
