```python
#3d array

v3 = [5,-4,3]

fig = plt.figure(figsize = plt.figaspect(1))
ax = plt.axes(projection = "3d")
ax.plot([0, v3[0]], [0, v3[1]], [0, v3[2]], linewidth = 3)

ax.plot([0,0],[-5,5],[0,0], "k--")
ax.plot([0,0],[0,0],[-6,6], "k--")
ax.plot([-5,5],[0,0],[0,0], "k--")
plt.show()
```

```python
#vector operations

a = np.array([4,5,6])
b = np.array([10,20,30])
c = np.array({0,3,6,9})

a+b
#a+c = error
```

```python
#vector geometric operations

v1 = np.array([4, -2])
v2 = np.array([3, 5])
v3 = v1+v2
print(v3)

#plotting 2d
plt.plot([0, v1[0]], [0, v1[1]], "b", label = "v1")
plt.plot([0, v2[0]]+v1[0], [0, v2[1]]+v1[1], "r", label = "v2")
plt.plot([0, v3[0]], [0, v3[1]], "g", label = "v1+v2")

plt.plot([-7,7],[0,0],"k--")
plt.plot([0,0],[-7,7],"k--")

plt.legend()
plt.axis("square")
plt.axis((-8,8,-8,8))
plt.grid()
plt.show()
```
