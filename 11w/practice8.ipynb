# 10- Answer this question by using Bayes Theorem and the law of the Total Probaility


```python
P_A = 22 / 118
P_not_A = 96 / 118

# P(B|A) 
P_B_given_A = 3 / 22
# P(B|not A) 
P_B_given_not_A = 57 / 96

# Law of Total Probability for P(B) 
P_B_total = (P_B_given_A * P_A) + (P_B_given_not_A * P_not_A)

# Bayes Theorem
P_A_given_B = (P_B_given_A * P_A) / P_B_total

print(round(P_A_given_B, 2))
```

# 11- Answer this question by using Bayes Theorem only.
