Python lists are generic containers, not mathematical vectors. In Python, the * operator applied to a list and an integer performs list repetition (concatenation). Instead of scaling the values, it simply repeats the sequence four times.

```python
v = np.array(v)
v2 = v*k
v2
```

```python
#Q2
a = np.array([5, -3])
lmbd = -2
a_scaled = lmbd * a
print(a_scaled)

#magnitude
mag_a = np.linalg.norm(a)
mag_scaled = np.linalg.norm(a_scaled)

print(mag_a) #magnitude of the vector
print(mag_scaled) #magnitude of new vector
print(mag_scaled / mag_a)
```

The magnitude increased by a factor of 2 (absolute value of lambda)

The direction was reversed (180-degree flip) because lambda is negative

NumPy allows this through a mechanism called broadcasting.

The scalar is treated as an array of the same shape as vector v, and the addition is performed element-wise.
    
This is done for efficiency and to simplify mathematical operations on datasets.

```python
#Q5

a = np.array([1, 2, 3, 4, 5, 6])

#Euclidean norm
squares = a**2
sum_of_squares = np.sum(squares)
manual_norm = np.sqrt(sum_of_squares)
print(manual_norm)

numpy_norm = np.linalg.norm(a)
print(numpy_norm)
```

```python
#Q6

v = np.random.rand(10)
norm_before = np.linalg.norm(v)
v_normalized = v / norm_before

norm_after = np.linalg.norm(v_normalized)

print(v)
print(norm_before)
print(norm_after)
```

```python
#Q7
a = np.array([1, 2, 3, 4, 5, 6])
b = np.array([0, -4, 3, -3, 6, 5])

manual_dot = 0
for i in range(len(a)):
    manual_dot += a[i] * b[i]

print(manual_dot)


numpy_dot = np.dot(a, b)
print(numpy_dot)
```

```python
#Q8

a = np.random.randn(3)
b = np.random.randn(3)
c = np.random.randn(3)

left_side = np.dot(a, (b + c))

right_side = np.dot(a, b) + np.dot(a, c)

print(a)
print(b)
print(c)

print(left_side)
print(right_side)

is_verified = np.allclose(left_side, right_side)
print(is_verified)
```

```python
#Q9
A = np.array([
    [2, 1],
    [5, 4],
    [1, 6],
    [7, 2]
])

B = np.array([
    [0, 0],
    [6, 3],
    [3, 5]
])


diff = A[:, np.newaxis, :] - B
dist_matrix = np.sqrt(np.sum(diff**2, axis=2))
print(dist_matrix.round(2))
```
