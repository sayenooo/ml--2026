```python
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

```python
a = np.array([1,2,3,4,5,6])

# methods 1, just like with the regular dot product
al1 = np.sqrt(sum(np.multiply(a,a)))

# method 2: take the norm
al2 = np.linalg.norm(a)

print(al1)
print(al2)
```

```python
#The dot product ( Vector-Vector multiplication)

#many ways to compute the dot product
a = np.array([1,2,3,4,5,6])
b = np.array([0,-4,3,-3,6,5])

#Option 1
ab1 = 0
for i in range(len(a)):
    # multiply corresponding element and sum
    ab1 = ab1 + a[i]*b[i]

#Option 2
ab2 = np.dot(a,b)

#Option 3
ab3 = sum(np.multiply(a,b))

#Option 4
ab4 = np.matmul(a,b)

print(ab1, ab2, ab3, ab4)
```

```python
#Dot product properties:

#Distributive

n = 10
a = np.random.randn(n)
b = np.random.randn(n)
c = np.random.randn(n)

r1 = np.dot(a,b) + np.dot(a,c)
r2 = np.dot( a,(b+c))

print(r1,r2)
```

```python
#Associative

#Special cases where associative property defined
#1. One vector is zero vector
#2. a=b=c

r1 = np.dot(a, np.dot(b,c))
r2 = np.dot(np.dot(b,c), a)

print(r1,r2)
```

```python
#The Geometric perspective of the dot product

a = np.array([1,3,-2])
b = np.array([1,-4,-4])

# compute the angle (radians) between two vectors
ang = np.arccos( np.dot(a,b)/(np.linalg.norm(a)*np.linalg.norm(b)) )
ang_deg = np.round((np.degrees(ang)), 2)

fig = plt.figure()
ax = plt.axes(projection = "3d")

ax.plot([0,a[0]],[0,a[1]],[0,a[2]], "k")
ax.plot([0,b[0]],[0,b[1]],[0,b[2]], "r")

plt.axis((-5,5,-5,5,-5,5))
plt.title('The angle betweenvectors a and b: %s deg.' %ang_deg)
plt.show()
```

```python
#Check if the algebraic and geometric formulas of dot product are the same or not

ab_a = np.dot(a,b)

ab_g = np.linalg.norm(a)*np.linalg.norm(b)*np.cos(ang)

print(ab_a)
print(ab_g)
```

```python
#Hadamard multiplication

a = [1,3,5]
b = [3,4,2]

ab_h = np.multiply(a,b)
print(ab_h)
```

```python
#Outer product

op = np.outer(a,b)
print(op)
```

```python
#Unit vector

a = np.array([-2,5])

sca = 1/np.linalg.norm(a)

a_sca = a*sca

plt.plot([0,a_sca[0]],[0,a_sca[1]], "k--", label = "||a||", linewidth = 5)
plt.plot([0,a[0]],[0,a[1]], "b", label = "a")


#OPTIONAL PART

plt.plot([-6,6],[0,0], "k--")
plt.plot([0,0],[-6,6], "k--")

plt.axis("square")
plt.axis((-6,6,-6,6))
plt.grid()
plt.legend()
plt.show()
```
